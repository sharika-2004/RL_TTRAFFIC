```python
import numpy as np
import random
```

Basic Traffic Environment

```python
class TrafficEnvironment:
    def __init__(self):
        self.lanes = [0, 0, 0, 0]  # 4 lanes
        self.current_green = 0
        
    def reset(self):
        self.lanes = [random.randint(5, 20) for _ in range(4)]
        self.current_green = 0
        return self.lanes
    
    def step(self, action):
        self.current_green = action
        
        # Vehicles pass in green lane
        self.lanes[action] = max(0, self.lanes[action] - random.randint(5, 10))
        
        # New vehicles arrive
        for i in range(4):
            self.lanes[i] += random.randint(0, 5)
        
        reward = -sum(self.lanes)
        
        return self.lanes, reward
```

Run Simulation

```python
env = TrafficEnvironment()
lanes = env.reset()

lane_names = ["North", "South", "East", "West"]

for step in range(10):
    action = random.randint(0, 3)
    
    lanes, reward = env.step(action)
    
    print(f"Step {step+1}")
    print(f"Green Lane: {lane_names[action]}")
    print(f"Lanes: {lanes}")
    print(f"Reward: {reward}")
    print("-"*30)
```

```python
import numpy as np
import random

class TrafficEnvironment:
    def __init__(self):
        self.lanes = [0, 0, 0, 0]
        self.current_green = 0
        self.time_step = 0
        
    def reset(self):
        self.lanes = [random.randint(10, 30) for _ in range(4)]
        self.current_green = 0
        self.time_step = 0
        return self.lanes
    
    def step(self, action):
        self.current_green = action
        self.time_step += 1
        
        # 🚦 Vehicles pass in green lane
        self.lanes[action] = max(0, self.lanes[action] - random.randint(8, 15))
        
        # 🚗 Traffic arrival logic
        for i in range(4):
            
            # Rush hour every 5 steps
            if self.time_step % 5 == 0:
                incoming = random.randint(5, 10)
            else:
                incoming = random.randint(1, 5)
            
            # Lane imbalance
            if i == random.randint(0, 3):
                incoming += random.randint(3, 7)
            
            self.lanes[i] += incoming
        
        # 🎯 Improved reward
        reward = -sum(self.lanes) - (max(self.lanes) - min(self.lanes))
        
        return self.lanes, reward
```

```python
env = TrafficEnvironment()
lanes = env.reset()

lane_names = ["North", "South", "East", "West"]

for step in range(10):
    action = random.randint(0, 3)
    
    lanes, reward = env.step(action)
    
    print(f"Step {step+1}")
    print(f"Green Lane: {lane_names[action]}")
    print(f"Lanes: {lanes}")
    print(f"Reward: {reward}")
    print("-"*30)
```

The environment was enhanced by incorporating time-dependent traffic patterns and lane imbalance to simulate realistic urban traffic conditions.
